```python
movies_df = pd.read_csv('data/movies.csv')
movies_df.info()
movies_df.head()
```

```python
# Xử lý DL cho SVD
# Đọc dữ liệu ratings, dùng copy không pandas sẽ sinh ra cái gì đó? 
ratings_svd = ratings_df[
    [
        "userId",
        "movieId",
        "rating"
    ]
].copy()
ratings_svd.info()
ratings_svd.head()
```

```python
# kiểm tra xem có bị ngoài khoảng (0.5 - 5)?
ratings_svd["rating"].describe()
```

```python
# kiểm tra số lượng User
ratings_svd["userId"].nunique()
```

```python
ratings_svd["movieId"].nunique()
```

```python
# Tạo cột year từ title
movies_df["year"] = movies_df["title"].str.extract(r"\((\d{4})\)")
```

```python
# Chuyển về số, lỗi sẽ thành NaN
movies_df["year"] = pd.to_numeric(
    movies_df["year"],
    errors="coerce"
)
movies_df.head()
```

```python
# Xóa năm khỏi title
movies_df["title"] = movies_df["title"].str.replace(
    r"\(\d{4}\)",
    "",
    regex=True
).str.strip()
movies_df.head()
```

```python
# Tính sparsity để xem độ thưa là bao nhiêu, và nêu lý do chọn SVD và SVD sẽ xử lý tốt dữ liệu thưa (không bắt buộc, chỉ tính để biết nó thưa và SVD làm tốt với DL thưa)
n_users = ratings_svd["userId"].nunique()
n_movies = ratings_svd["movieId"].nunique()
n_ratings = len(ratings_svd)

sparsity = (
    1 - n_ratings / (n_users * n_movies)
) * 100

print(f"Sparsity: {sparsity:.2f}%")
```

```python
# Xử lý DL cho NPL
# vì ở trên đã kiểm tra và thấy bảng movies không có dữ liệu null, nhưng lại có phim không được gán nhãn nhưng vẫn có giá trị (no genres listed) - không có ý nghĩa
movies_df["genres"].unique()
```

```python
# Đếm có bao nhiêu phim không được gán nhãn, được gán giá trị (no genres listed)
count = len(
    movies_df[movies_df["genres"] == "(no genres listed)"]
)
print(count)
```

```python
# Thay (no genres listed) thành ""
movies_df["genres"] = movies_df["genres"].replace(
    "(no genres listed)",
    ""
)
```

```python
# Kiểm tra lại đã thay chưa
count = len(
    movies_df[movies_df["genres"] == "(no genres listed)"]
)
print(count)
```

```python
# Thay dấu | trong genres bằng khoảng trắng
movies_df["genres"] = movies_df["genres"].str.replace(
    "|",
    " ",
    regex=False
)
movies_df.head()
```

```python
# Xử lý tags
# Chuẩn hóa thành chữ thường 
tags_df["tag"] = tags_df["tag"].str.lower()
tags_df.head()
```

```python
# Loại bỏ khoảng trắng thừa, ở 2 đầu nếu người dùng lỡ bấm space trước/sau khi viết 
tags_df["tag"] = tags_df["tag"].str.strip()
tags_df.head()
```

```python
# Gom tag theo movie
movie_tags = tags_df.groupby("movieId")["tag"].apply(" ".join).reset_index()
movie_tags.head()
```

```python
# Ghép tag vào movies
movies_nlp = movies_df.merge(
    movie_tags,
    on="movieId",
    how="left"
)
movies_nlp.head()
```

```python
# Xử lý Null
movies_nlp["tag"] = (
    movies_nlp["tag"]
    .fillna("")
)
movies_nlp.head()
```

```python
# Ghép title, genres, tag thành 1 cột content 
movies_nlp["content"] = (
    movies_nlp["title"]
    + " "
    + movies_nlp["genres"]
    + " "
    + movies_nlp["tag"]
)
movies_nlp.head()
```

```python
# Chuyển về chữ thường 
movies_nlp["content"] = (
    movies_nlp["content"]
    .str.lower()
)
movies_nlp.head()
```

```python
# Đếm số ký tự đặc biệt trong content 
special_content = movies_nlp[
    movies_nlp["content"].str.contains(
        r"[^a-zA-Z0-9\s]",
        regex=True,
        na=False
    )
]

print(len(special_content))
```

```python
# Loại bỏ ký tự đặc biệt, chỉ giữ lại chữ cái, số và khoảng trắng
movies_nlp["content"] = (
    movies_nlp["content"]
    .str.replace(
        r"[^a-zA-Z0-9\s]",
        " ",
        regex=True
    )
)
movies_nlp.head()
```

```python
# Loại bỏ khoảng trắng thừa, ở giữa nếu có nhiều khoảng trắng sẽ thành 1 khoảng trắng, ở 2 đầu nếu có sẽ bị xóa
movies_nlp["content"] = (
    movies_nlp["content"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
movies_nlp.head()
```

```python
# Tính TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(
    stop_words="english"
)
tfidf_matrix = tfidf.fit_transform(
    movies_nlp["content"]
)
print(tfidf_matrix.shape)
```

```python
feature_names = tfidf.get_feature_names_out()

print(feature_names[:200])
```

```python
# Tính cosine similarity để đo độ tương đồng giữa các vector phim.
from sklearn.metrics.pairwise import cosine_similarity
cosine_sim = cosine_similarity(
    tfidf_matrix,
    tfidf_matrix
)
print(cosine_sim.shape)
```

```python
# Tạo Series để map index với title, dùng drop_duplicates để tránh trường hợp có phim trùng tên
indices = pd.Series(
    movies_nlp.index,
    index=movies_nlp["title"]
).drop_duplicates()
indices.head()
```

```python
# Hàm đề xuất phim dựa trên title, trả về 10 phim tương tự nhất
def recommend_movies(title, cosine_sim=cosine_sim): 
    idx = indices[title]
    sim_scores = list(
        enumerate(cosine_sim[idx])
    )
    sim_scores = sorted(
        sim_scores,
        key=lambda x: x[1],
        reverse=True
    )
    sim_scores = sim_scores[1:11]

    movie_indices = [
        i[0] for i in sim_scores
    ]
    return movies_nlp[
        ["title","genres"]
    ].iloc[movie_indices]
```

```python
# Test hàm đề xuất
recommend_movies("Toy Story")
```

```python
# =====================================================================
# NHÁNH 2: TRÍCH CHỌN ĐẶC TRƯNG ẨN SVD (HÀNH VI NGƯỜI DÙNG)
# =====================================================================

from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split


# 1. Định dạng cấu trúc dữ liệu tương tác cho thư viện Surprise từ bảng ratings_svd ở Cell 6
reader = Reader(rating_scale=(0.5, 5.0))
data_surprise = Dataset.load_from_df(ratings_svd, reader)


# 2. Chia dữ liệu thành 80% Train (huấn luyện) và 20% Test (kiểm thử)
trainset, testset = train_test_split(data_surprise, test_size=0.20, random_state=42)


# 3. Khởi tạo thuật toán Funk SVD và tiến hành huấn luyện (fit) để học các đặc trưng ẩn
svd_model = SVD(n_factors=100, random_state=42)
svd_model.fit(trainset)


print("--- KẾT QUẢ TRÍCH CHỌN ĐẶC TRƯNG SVD ---")
# 4. Xuất kích thước và xem thử ma trận đặc trưng ẩn của Người dùng (P)
user_features = svd_model.pu
print(f"Kích thước ma trận đặc trưng ẩn Người dùng (P): {user_features.shape}")
print(f"Vector đặc trưng ẩn của User đầu tiên (10 chiều đầu): \n{user_features[0][:10]}\n")


# 5. Xuất kích thước và xem thử ma trận đặc trưng ẩn của Bộ phim (Q)
movie_features = svd_model.qi
print(f"Kích thước ma trận đặc trưng ẩn Bộ phim (Q): {movie_features.shape}")
print(f"Vector đặc trưng ẩn của Movie đầu tiên (10 chiều đầu): \n{movie_features[0][:10]}")
```

Ý tưởng chính:

SVD dự đoán user có thể chấm một phim bao nhiêu điểm.
TF-IDF biểu diễn nội dung phim bằng vector số.
User Profile được tạo từ các phim user đã đánh giá cao.
HybridScore kết hợp điểm SVD và điểm nội dung để xếp hạng phim chưa xem.

```python
# Kiểm tra lại các biến quan trọng đã có từ phần trên
required_variables = [
    "ratings_svd",
    "movies_nlp",
    "tfidf_matrix",
    "svd_model",
    "testset"
]

for var_name in required_variables:
    if var_name not in globals():
        raise NameError(f"Bạn cần chạy các cell phía trên trước. Biến còn thiếu: {var_name}")

print("Các biến cần thiết đã sẵn sàng.")
print("ratings_svd:", ratings_svd.shape)
print("movies_nlp:", movies_nlp.shape)
print("tfidf_matrix:", tfidf_matrix.shape)
```

Đánh giá mô hình SVD đã huấn luyện
Ở cell trước, mô hình SVD đã được huấn luyện bằng lệnh svd_model.fit(trainset). Bước này dùng tập testset đã tách ở trên để kiểm tra mô hình dự đoán rating tốt hay chưa.

Hai chỉ số thường dùng:

RMSE: sai số căn bậc hai trung bình, phạt mạnh các lỗi lớn.
MAE: sai số tuyệt đối trung bình.
Hai chỉ số này càng thấp thì mô hình dự đoán rating càng tốt.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import pandas as pd


def evaluate_svd_model(model, testset):
    y_true = []
    y_pred = []

    for user_id, movie_id, true_rating in testset:
        predicted_rating = model.predict(user_id, movie_id).est

        y_true.append(true_rating)
        y_pred.append(predicted_rating)

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)

    return rmse, mae


svd_rmse, svd_mae = evaluate_svd_model(svd_model, testset)

print("SVD RMSE:", svd_rmse)
print("SVD MAE:", svd_mae)
```

Xem user thích những thể loại nào
Bước này chưa phải gợi ý phim, mà dùng để giải thích hệ thống hiểu sở thích người dùng như thế nào.

Cách làm:

Lấy các phim mà user đã đánh giá cao, ví dụ rating từ 3.5 trở lên.
Ghép với bảng phim để lấy genres.
Đếm thể loại nào xuất hiện nhiều và có rating trung bình cao.
Kết quả giúp giải thích rằng hệ thống không cần người dùng nhập tên phim, mà tự học gu từ lịch sử rating.

```python
def get_user_favorite_genres(user_id, ratings_df, movies_df, min_rating=3.5):
    user_ratings = ratings_df[
        (ratings_df["userId"] == user_id) &
        (ratings_df["rating"] >= min_rating)
    ].copy()

    if user_ratings.empty:
        return pd.DataFrame(columns=["genre", "count", "avg_rating"])

    user_movies = user_ratings.merge(
        movies_df[["movieId", "genres"]],
        on="movieId",
        how="left"
    )

    genre_stats = {}

    for row in user_movies.itertuples():
        genres_text = str(row.genres).replace("|", " ")
        genres = genres_text.split()

        for genre in genres:
            if genre not in genre_stats:
                genre_stats[genre] = {
                    "total_rating": 0,
                    "count": 0
                }

            genre_stats[genre]["total_rating"] += row.rating
            genre_stats[genre]["count"] += 1

    result = []

    for genre, stats in genre_stats.items():
        result.append({
            "genre": genre,
            "count": stats["count"],
            "avg_rating": stats["total_rating"] / stats["count"]
        })

    return (
        pd.DataFrame(result)
        .sort_values(by=["avg_rating", "count"], ascending=False)
        .reset_index(drop=True)
    )


# Ví dụ: xem gu thể loại của user 1
get_user_favorite_genres(
    user_id=1,
    ratings_df=ratings_svd,
    movies_df=movies_nlp,
    min_rating=3.5
).head(10)
```

Tạo ánh xạ từ movieId sang vị trí trong ma trận TF-IDF
Trong bảng rating, mỗi phim được nhận diện bằng movieId.

Nhưng trong tfidf_matrix, mỗi phim được nhận diện bằng index dòng: 0, 1, 2, ...

Vì vậy cần tạo bảng tra cứu:

movieId → index trong tfidf_matrix

```python
movie_id_to_index = pd.Series(
    movies_nlp.index,
    index=movies_nlp["movieId"]
).to_dict()

print("Số movieId map được:", len(movie_id_to_index))
```

Xây dựng hồ sơ sở thích người dùng bằng TF-IDF
Đây là bước quan trọng của phần Content-based cá nhân hóa.

Thay vì nhập một tên phim rồi tìm phim tương tự, ta làm như sau:

Lấy các phim user đã rating cao.
Lấy vector TF-IDF của các phim đó.
Tính trung bình có trọng số theo rating.
Phim user chấm 5 sao sẽ ảnh hưởng mạnh hơn phim user chấm 3.5 sao.

Kết quả là một vector gọi là user_profile, biểu diễn gu nội dung của user.

```python
from scipy.sparse import vstack


def build_user_profile(user_id, ratings_df, min_rating=3.5):
    user_ratings = ratings_df[
        (ratings_df["userId"] == user_id) &
        (ratings_df["rating"] >= min_rating)
    ].copy()

    if user_ratings.empty:
        return None

    movie_vectors = []
    weights = []

    for row in user_ratings.itertuples():
        movie_id = row.movieId

        if movie_id in movie_id_to_index:
            movie_index = movie_id_to_index[movie_id]
            movie_vectors.append(tfidf_matrix[movie_index])
            weights.append(row.rating)

    if len(movie_vectors) == 0:
        return None

    movie_matrix = vstack(movie_vectors)
    weights = np.array(weights)

    # Nhân mỗi vector phim với rating tương ứng, sau đó lấy trung bình có trọng số
    user_profile = movie_matrix.multiply(weights[:, None]).sum(axis=0) / weights.sum()

    return np.asarray(user_profile)


user_profile_demo = build_user_profile(
    user_id=1,
    ratings_df=ratings_svd,
    min_rating=3.5
)

print("Kích thước user profile:", user_profile_demo.shape)
```

Tính ContentScore cho từng phim
ContentScore trả lời câu hỏi:

Phim này có giống gu nội dung của user không?

Cách tính:

So sánh vector user_profile với vector TF-IDF của toàn bộ phim.
Dùng Cosine Similarity để lấy điểm từ 0 đến 1.
Điểm càng cao thì phim càng giống sở thích nội dung của user.

```python
def get_user_content_scores(user_id, ratings_df, min_rating=3.5):
    user_profile = build_user_profile(
        user_id=user_id,
        ratings_df=ratings_df,
        min_rating=min_rating
    )

    if user_profile is None:
        return None

    content_scores = cosine_similarity(
        user_profile,
        tfidf_matrix
    ).flatten()

    return content_scores


content_scores_demo = get_user_content_scores(
    user_id=1,
    ratings_df=ratings_svd,
    min_rating=3.5
)

print("Số điểm ContentScore của user 1:", len(content_scores_demo))
print("5 điểm đầu:", content_scores_demo[:5])
```

Tạo hàm gợi ý phim Hybrid theo userId
Bây giờ kết hợp hai nguồn điểm:

SVDScore: điểm rating mà SVD dự đoán user sẽ chấm cho phim.
ContentScore: điểm phim đó giống gu nội dung của user.
Vì SVD dự đoán rating trong khoảng 0.5 đến 5.0, còn ContentScore nằm trong khoảng 0 đến 1, ta chuẩn hóa điểm SVD về 0 đến 1.

Công thức:

HybridScore = alpha * SVDScore_norm + (1 - alpha) * ContentScore

Với alpha = 0.7, hệ thống ưu tiên 70% hành vi rating và 30% nội dung phim.

```python
def normalize_svd_score(score, min_rating=0.5, max_rating=5.0):
    return (score - min_rating) / (max_rating - min_rating)


def recommend_movies_for_user(user_id, top_n=10, alpha=0.7, min_rating_for_profile=3.5):
    content_scores = get_user_content_scores(
        user_id=user_id,
        ratings_df=ratings_svd,
        min_rating=min_rating_for_profile
    )

    if content_scores is None:
        return pd.DataFrame({
            "message": [f"User {user_id} chưa có đủ phim rating cao để tạo hồ sơ sở thích."]
        })

    watched_movie_ids = set(
        ratings_svd[ratings_svd["userId"] == user_id]["movieId"]
    )

    recommendations = []

    for movie_index, movie in movies_nlp.iterrows():
        movie_id = movie["movieId"]

        # Không gợi ý lại phim user đã xem/rating
        if movie_id in watched_movie_ids:
            continue

        svd_pred_rating = svd_model.predict(user_id, movie_id).est
        svd_score_norm = normalize_svd_score(svd_pred_rating)
        content_score = content_scores[movie_index]

        hybrid_score = alpha * svd_score_norm + (1 - alpha) * content_score

        recommendations.append({
            "movieId": movie_id,
            "title": movie["title"],
            "genres": movie["genres"],
            "svd_pred_rating": svd_pred_rating,
            "content_score": content_score,
            "hybrid_score": hybrid_score
        })

    recommendations_df = pd.DataFrame(recommendations)

    return (
        recommendations_df
        .sort_values(by="hybrid_score", ascending=False)
        .head(top_n)
        .reset_index(drop=True)
    )
```

Chạy thử hệ thống gợi ý cá nhân hóa
Ở bước này chỉ cần nhập userId.

Hệ thống sẽ:

Hiển thị các thể loại user có xu hướng thích.
Gợi ý top phim user chưa xem dựa trên HybridScore.

```python
demo_user_id = 1

print("Các thể loại user có xu hướng thích:")
display(
    get_user_favorite_genres(
        user_id=demo_user_id,
        ratings_df=ratings_svd,
        movies_df=movies_nlp,
        min_rating=3.5
    ).head(10)
)

print("Top phim gợi ý theo Hybrid:")
display(
    recommend_movies_for_user(
        user_id=demo_user_id,
        top_n=10,
        alpha=0.7,
        min_rating_for_profile=3.5
    )
)
```

Thử nhiều giá trị alpha
Bước này giúp giải thích phần hyper-parameter của mô hình Hybrid.

alpha càng lớn: hệ thống dựa nhiều hơn vào SVD.
alpha càng nhỏ: hệ thống dựa nhiều hơn vào nội dung phim.
Cell này không dùng để đánh giá RMSE trực tiếp, mà dùng để quan sát danh sách gợi ý thay đổi như thế nào khi thay đổi trọng số Hybrid.

```python
for alpha in [0.5, 0.7, 0.9]:
    print(f"
Top 5 phim gợi ý cho user {demo_user_id} với alpha = {alpha}")
    display(
        recommend_movies_for_user(
            user_id=demo_user_id,
            top_n=5,
            alpha=alpha,
            min_rating_for_profile=3.5
        )[["title", "genres", "svd_pred_rating", "content_score", "hybrid_score"]]
    )
```
